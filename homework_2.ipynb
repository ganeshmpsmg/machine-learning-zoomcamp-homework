```python
n_train = int(n * 0.6)
n_val = int(n * 0.2)
n_test = n - n_train - n_val

n_train, n_val, n_test
```

```python
df_train = df.iloc[:n_train]
df_val = df.iloc[n_train:n_train+n_val]
df_test = df.iloc[n_train+n_val:]
```

```python
y_train = np.log1p(df_train.msrp.values)
y_val = np.log1p(df_val.msrp.values)
y_test = np.log1p(df_test.msrp.values)
```

```python
def train_linear_regression(X, y):
    ones = np.ones(X.shape[0])
    
    X = np.column_stack([ones, X])
    
    XTX = X.T.dot(X)
    XTX_inv = np.linalg.inv(XTX)
    
    w = XTX_inv.dot(X.T).dot(y)
    
    return w
```

```python
def predict_linear_regression(X, w):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])
    
    return X.dot(w)
```

```python
base_features = [
    'engine_hp',
    'engine_cylinders',
    'highway_mpg',
    'city_mpg',
    'popularity'
]
```

```python
def prepare_X(df, features):
    df = df.copy()
    
    X = df[features].copy()
    
    X = X.fillna(0)
    
    return X.values
```

```python
X_train = prepare_X(df_train, base_features)

X_train.shape
```

```python
w = train_linear_regression(X_train, y_train)

w
```

```python
y_pred = predict_linear_regression(X_train, w)

y_pred[:10]
```

```python
def train_linear_regression(X, y):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)
    XTX_inv = np.linalg.inv(XTX)

    w = XTX_inv.dot(X.T).dot(y)

    return w
```

```python
def rmse(y, y_pred):
    error = y_pred - y
    
    mse = np.mean(error ** 2)
    
    return np.sqrt(mse)
```

```python
rmse(y_train, y_pred)
```

```python
categorical = [
    'make',
    'model',
    'transmission_type',
    'driven_wheels',
    'vehicle_size',
    'vehicle_style'
]
```

```python
df_train = pd.get_dummies(
    df_train,
    columns=categorical,
    dtype=int
)

df_val = pd.get_dummies(
    df_val,
    columns=categorical,
    dtype=int
)
```

```python
def train_linear_regression_reg(X, y, r=0.00):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)

    reg = r * np.eye(XTX.shape[0])
    reg[0, 0] = 0

    XTX_inv = np.linalg.inv(XTX + reg)

    w = XTX_inv.dot(X.T).dot(y)

    return w
```

```python
for r in [0, 0.01, 0.1, 1, 10, 100]:
    
    w = train_linear_regression_reg(
        X_train,
        y_train,
        r=r
    )
    
    y_pred = predict_linear_regression(
        X_val,
        w
    )
    
    print(r, rmse(y_val, y_pred))
```

```python
for r in [0, 0.000001, 0.00001, 0.0001, 0.001, 0.01, 0.1, 1]:
    
    w = train_linear_regression_reg(
        X_train,
        y_train,
        r=r
    )
    
    y_pred = predict_linear_regression(
        X_val,
        w
    )
    
    score = rmse(y_val, y_pred)
    
    print(f"r={r}, RMSE={score}")
```

```python
df_full_train = pd.concat(
    [df_train, df_val]
).reset_index(drop=True)

y_full_train = np.concatenate(
    [y_train, y_val]
)
```

```python
X_full_train = prepare_X(
    df_full_train,
    features
)

X_test = prepare_X(
    df_test,
    features
)
```

```python
r = 0.01

w = train_linear_regression_reg(
    X_full_train,
    y_full_train,
    r=r
)
```

```python
y_pred_test = predict_linear_regression(
    X_test,
    w
)
```

```python
rmse(
    y_test,
    y_pred_test
)
```
